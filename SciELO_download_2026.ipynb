# SciELO: download do dump (2026)

Baixa o arquivo `scielo_articles.zip`, o mesmo usado no projeto insyspo/scielo, e registra data, tamanho e SHA-256 para sabermos exatamente de qual versão do dump vieram os dados.

```python
import os
import requests
```

```python
ZIP_URL = 'http://static.scielo.org/articlemeta/scielo_articles.zip'
ZIP_PATH = 'data/scielo_2026/scielo_articles.zip'
```

```python
os.makedirs('data/scielo_2026', exist_ok=True)
```

```python
if os.path.exists(ZIP_PATH):
    print('Already downloaded:', ZIP_PATH)
else:
    with requests.get(ZIP_URL, stream=True) as r:
        r.raise_for_status()
        with open(ZIP_PATH, 'wb') as f:
            for chunk in r.iter_content(8 * 1024 * 1024):
                f.write(chunk)

print(os.path.getsize(ZIP_PATH) / 1e9, 'GB')
```
